```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import ( accuracy_score, precision_score, recall_score,f1_score, roc_auc_score, roc_curve, confusion_matrix)
from sklearn import tree
```

We will use the get_dummies() function to convert object and category datatypes to dummy variables

```python
# convert categorical variables to numeric  datatype

x_cat_features=['international plan', 'voice mail plan','area code']
X_cat= pd.get_dummies(Syriatel_data[x_cat_features], drop_first =True, dtype= float)
#The drop_first=True argument removes the first level for each categorical variable 
#the dtype=float argument converts the datatype of all the dummy variables to float
```

```python
X= pd.get_dummies(Syriatel_data.drop(columns=['churn','state','total_individual_charges', 'total_individual_minutes','total_individual_calls', 'charge_per_call']), drop_first=True, dtype=float)
X .head()   
# we have dropped the target variable, state and the other columns which have a  high correlation with each other
                  
```

*<span style="font-size: 20px;">5.1.2 Performing a train test split </span>*

```python

X_train,X_test, y_train, y_test= train_test_split(X,y, test_size= 0.3, random_state=42, stratify= y)
#stratify= y -to keep the classes in churn balanced for X_train and y_train

# standardizing
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```python
print('Train shape:', X_train.shape, 'Test shape:', X_test.shape)
```

### Train Baseline Models

We begin by training two baseline models to predict customer churn:

- **Logistic Regression**:  
  A simple and interpretable model that provides probability estimates. It serves as a strong baseline for comparison and helps understand the relationship between features and churn.

- **Decision Tree**:  
  A non-linear model that is easy to interpret and visualize. It captures complex patterns in the data and helps identify key decision rules driving churn.

These baseline models will later be enhanced using a pipeline and compared with an ensemble model (Random Forest) to improve performance.

```python
y_train.value_counts()
```

We're going to fix class imbalance with both stratified sampling and smote

Since there is about 80% non churners and 15% churners there is class imbalance in our dataset which will also cause imbalance in our training and test set.

To fix this we use Smote to generate synthetic samples for the minority class.


```python
from imblearn.over_sampling import SMOTE
import numpy as np
```

```python
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_train, y_train)
```

```python
print('resampled train shape:', X_resampled.shape)
```

```python
y_resampled.value_counts()
```

We can now see that our classes have been balanced. 

*<span style="font-size: 20px;">5.1.3 Fitting a logistic model  </span>*

```python
log_model = LogisticRegression(max_iter=1000, random_state=42)
log_model.fit(X_resampled, y_resampled)
y_pred_log = log_model.predict(X_test)
y_proba_log = log_model.predict_proba(X_test)[:, 1]  # probability of churn
#the y_pred_proba gives us the probability of the true churns only
log_model

```

*<span style="font-size: 20px;">5.1.4 Fitting a Classifier model for Decision Tree </span>*

```python
# Decision Tree
clf_model = DecisionTreeClassifier(random_state=42)
clf_model.fit(X_resampled, y_resampled)
y_pred_dt = clf_model.predict(X_test)
# decision tree probabilities (for ROC/AUC)
y_proba_dt = clf_model.predict_proba(X_test)[:, 1]

clf_model

```

```python
## plotting the decision tree
fig, axes = plt.subplots(nrows = 1,ncols = 1, figsize = (3,3), dpi=1000)

tree.plot_tree(clf_model,
               feature_names = X.columns, 
               class_names=np.unique(y).astype('str'),
               filled = True)
plt.show()
```

## Observation
* Our leaf node has a gini impurity of 0 meaning there is high purity in the final subset.

## Model Improvements

To improve performance and ensure a more robust workflow, we introduce:

- A Pipeline to streamline preprocessing and modeling  
- A Random Forest ensemble model to improve predictive accuracy

*<span style="font-size: 20px;">5.1.5 Pipeline Implementation (Logistic Regression)  </span>*

A pipeline is used to combine preprocessing and model training into a single workflow. 
This ensures consistent data transformation and helps prevent data leakage.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# create pipeline
pipe_log = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=1000))
])

# train
pipe_log.fit(X_resampled, y_resampled)

# predict
y_pred_pipe = pipe_log.predict(X_test)
y_proba_pipe = pipe_log.predict_proba(X_test)[:, 1]
```

```python
# Evaluate the pipeline model
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

acc_pipe = accuracy_score(y_test, y_pred_pipe)
prec_pipe = precision_score(y_test, y_pred_pipe, zero_division=0)
rec_pipe = recall_score(y_test, y_pred_pipe, zero_division=0)
f1_pipe = f1_score(y_test, y_pred_pipe, zero_division=0)
auc_pipe = roc_auc_score(y_test, y_proba_pipe)
```

### Pipeline Results

The pipeline simplifies the workflow by combining scaling and modeling into one step. 
It produces similar or improved performance compared to the standalone Logistic Regression model while ensuring consistency and reducing the risk of errors.

*<span style="font-size: 20px;">5.1.6 Ensemble Model: Random Forest  </span>*

Random Forest is an ensemble method that combines multiple decision trees to improve prediction accuracy and reduce overfitting.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42)

rf_model.fit(X_resampled, y_resampled)

y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]
```

```python
# Evaluate the random forest model
acc_rf = accuracy_score(y_test, y_pred_rf)
prec_rf = precision_score(y_test, y_pred_rf, zero_division=0)
rec_rf = recall_score(y_test, y_pred_rf, zero_division=0)
f1_rf = f1_score(y_test, y_pred_rf, zero_division=0)
auc_rf = roc_auc_score(y_test, y_proba_rf)
```

### Random Forest Insights

The Random Forest model improves performance by averaging multiple decision trees, making it more robust and less prone to overfitting.

It typically achieves better accuracy and overall performance compared to a single Decision Tree.

***<span style="font-size: 24px;">6.0 Model Evaluation</span>***

For this part we are going to check how the model performs using metrics such as:
*  Accuracy: how many were correct overall.
*  Precision: of those we said "will churn", how many really churned.
*  Recall: of those who really churned, how many did we find.
*  F1: balance between precision and recall.
*  AUC: how well the model ranks people from low to high risk


```python

# Logistic
acc_log = accuracy_score(y_test, y_pred_log)
prec_log = precision_score(y_test, y_pred_log, zero_division=0)
rec_log = recall_score(y_test, y_pred_log, zero_division=0)
f1_log = f1_score(y_test, y_pred_log, zero_division=0)
auc_log = roc_auc_score(y_test, y_proba_log)

# Decision Tree
acc_dt = accuracy_score(y_test, y_pred_dt)
prec_dt = precision_score(y_test, y_pred_dt, zero_division=0)
rec_dt = recall_score(y_test, y_pred_dt, zero_division=0)
f1_dt = f1_score(y_test, y_pred_dt, zero_division=0)
auc_dt = roc_auc_score(y_test, y_proba_dt)

# Show results in a tiny table

results = pd.DataFrame({
    'Model': ['Logistic', 'DecisionTree', 'Pipeline', 'RandomForest'],
    'Accuracy': [acc_log, acc_dt, acc_pipe, acc_rf],
    'Precision': [prec_log, prec_dt, prec_pipe, prec_rf],
    'Recall': [rec_log, rec_dt, rec_pipe, rec_rf],
    'F1': [f1_log, f1_dt, f1_pipe, f1_rf],
    'AUC': [auc_log, auc_dt, auc_pipe, auc_rf]
})

results = results.round(3)
results
```
